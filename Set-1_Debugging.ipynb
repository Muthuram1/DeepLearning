```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

data = pd.DataFrame({
    'Feature1': np.random.rand(100),
    'Feature2': np.random.rand(100),
    'Label': np.random.randint(0, 2, size=100)
})

X = data[['Feature1', 'Feature2']]
y = data['Label']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

model = Sequential()
model.add(Dense(16, activation='relu', input_shape=(2,)))   # Fixed: 2 features
model.add(Dense(8, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])


model.fit(X_train, y_train, epochs=20, batch_size=16, validation_split=0.2)


eval_loss, eval_accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {eval_accuracy}")


predictions = model.predict(X_test)
predicted_classes = (predictions > 0.5).astype(int)
print(predicted_classes[:10])
```

```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers

X_train = np.random.rand(100, 28, 28, 1)
y_train = np.random.randint(0, 10, size=(100,))

model = keras.Sequential([                                   # Fixed: Sequential
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),  # Fixed: activation
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs=5, batch_size=32)
```
